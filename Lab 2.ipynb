```python
from Crypto.Cipher import AES
import binascii


# decrypts AES-encrypted data in CBC mode
def decrypt_aes_cbc(key_hex, ciphertext_hex):
    key = bytes.fromhex(key_hex)
    ciphertext = bytes.fromhex(ciphertext_hex)
    
    iv = ciphertext[:16]
    actual_ciphertext = ciphertext[16:]
    
    cipher = AES.new(key, AES.MODE_CBC, iv)
    plaintext_padded = cipher.decrypt(actual_ciphertext)
    
    # Remove PKCS#7 padding
    padding_len = plaintext_padded[-1]
    plaintext = plaintext_padded[:-padding_len]
    
    return plaintext.decode('utf-8')
```

```python
from Crypto.Cipher import AES
from Crypto.Util import Counter


# decrypts an AES-encrypted message that uses CTR mode
def decrypt_aes_ctr(key_hex, ciphertext_hex):
    key = bytes.fromhex(key_hex)
    ciphertext = bytes.fromhex(ciphertext_hex)
    
    iv = ciphertext[:16]
    actual_ciphertext = ciphertext[16:]
    
    # Create counter using IV as nonce
    ctr = Counter.new(128, initial_value=int.from_bytes(iv, byteorder='big'))
    cipher = AES.new(key, AES.MODE_CTR, counter=ctr)
    
    plaintext = cipher.decrypt(actual_ciphertext)
    return plaintext.decode('utf-8')
```
